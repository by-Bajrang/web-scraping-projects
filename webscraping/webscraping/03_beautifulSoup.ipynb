```python
from bs4 import BeautifulSoup
```

```python
with open("htmls/page1.html") as f:
    content = f.read()
```

```python
soup = BeautifulSoup( content , "html.parser")
```

```python
h3s = soup.find_all("h3")
```

```python
articles = soup.select("article.product_pod")
```

```python
itmes = []
for article in articles:
    title = article.find("h3").find("a")["title"]
    # print(title)
    price = article.select_one("p.price_color")
    # print(price.text)
    rating_element = article.select_one("p.star-rating")
    rating = rating_element['class'][1]   # rating_element['class'] === arr samja kaya 
    itmes.append([title,price.text,rating])
```

```python
itmes
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(itmes , columns=["Book Tile","Price","Rating"])
```
